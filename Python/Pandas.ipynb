#### **1. What is pandas?**
Pandas is an open-source Python library designed for high-performance data manipulation, analysis, and cleaning. Built on top of the NumPy library, it acts as the backbone for most data science and machine learning workflows in Python.

**Core Data Structures**

* **Series:** A one-dimensional labeled array capable of holding any data type (integer, string, float, python objects, etc.). Think of it as a single column in a table.
* **DataFrame:** A two-dimensional, size-mutable, tabular data structure with labeled axes (rows and columns). It is the primary pandas object and functions like a highly powered Excel spreadsheet or SQL table.

**Why It Is Essential**

* **I/O Operations:** It seamlessly reads and writes data from multiple formats, including CSV, Excel, SQL databases, Parquet, and JSON.
* **Data Cleaning:** It provides built-in methods for handling missing data (`NaN`), dropping duplicates, and standardizing data types.
* **Data Transformation:** It allows for effortless slicing, filtering, reshaping, aggregating (groupby), and merging of disparate datasets.
* **Time Series Analysis:** It includes native tools for date range generation, frequency conversion, and moving window statistics.

Pandas bridges the gap between raw, unstructured data and the clean datasets required for training machine learning models. It streamlines the exploratory data analysis and feature engineering steps that typically consume the majority of a data science workflow.

#### **3. What are the main data structures in Pandas?**
Pandas relies on two foundational data structures: the **Series** (1-dimensional) and the **DataFrame** (2-dimensional). Every operation in Pandas is essentially manipulating one or both of these structures.

**1. Series (1-Dimensional)**
A Series is a one-dimensional array-like object capable of holding any single data type (integers, strings, floats, objects). It consists of two main components: the actual data array and an associated array of data labels, called its *index*.

**2. DataFrame (2-Dimensional)**
A DataFrame is a two-dimensional, size-mutable tabular structure with rows and columns. You can think of it as a dictionary of aligned Series objects that all share the same index. This is the most commonly used pandas object.

**Core Differences**  
| Feature | Series | DataFrame |
| --- | --- | --- |
| **Dimensions** | 1D (Array/List-like) | 2D (Table/Grid-like) |
| **Data Types** | Homogeneous (single data type per Series) | Heterogeneous (each column can be a different data type) |
| **Structure** | Single column of data with an Index | Multiple columns of data, sharing a single Index |
| **Size Mutability** | Fixed size (adding elements requires recreating) | Mutable (easily add or drop rows/columns) |

```python
import pandas as pd

# Creating a Series from a list with a custom string index
pm25_values = pd.Series(
    [45.2, 112.5, 89.0], 
    index=['Morning', 'Afternoon', 'Evening'], 
    name='PM2.5_Levels'
)

print(pm25_values)
print("\nAccessing by index label:", pm25_values['Afternoon'])
```

```python
import pandas as pd

# Creating a DataFrame from a dictionary
data = {
    'Date': ['2026-09-10', '2026-09-11', '2026-09-12'],
    'Station': ['Zone A', 'Zone B', 'Zone A'],
    'AQI': [85, 142, 90],
    'Status': ['Good', 'Poor', 'Satisfactory']
}

df = pd.DataFrame(data)

# Setting a specific column as the row index
df.set_index('Date', inplace=True)

display(df)

# Extracting a single column returns a Series
print("\nType of a single column:", type(df['AQI']))
```

#### **4. What is a Series?**
A **Series** is a one-dimensional, labeled array capable of holding any data type (integers, strings, floating-point numbers, Python objects). Think of it as a single column in a spreadsheet, where every row has a specific, immutable label.

**Key Characteristics**
* **The Index:** Unlike a standard Python list, every element in a Series has an explicit, customizable label (the index). If you don't provide one, Pandas assigns a default sequence (`0, 1, 2...`).
* **Data Homogeneity:** While it *can* hold mixed types (which forces the overall type to `object`), a Series is highly optimized when holding a single data type, allowing for memory efficiency and high-speed calculations.
* **Dual Behavior:** A Series acts like a hybrid between a Python dictionary (key-value pairs via the index) and a NumPy array (vectorized math operations).

```python
import pandas as pd

# 1. Creating a Series from a standard list (receives a default integer index)
cities = pd.Series(['Bengaluru', 'Kolkata', 'Chennai', 'Mumbai'])
print("Default Index Series:\n", cities)

# 2. Creating a Series from a dictionary (dictionary keys automatically become the index)
air_quality_dict = {'Indiranagar': 85, 'HSR Layout': 142, 'Koramangala': 110}
aqi_series = pd.Series(air_quality_dict, name='AQI_Readings')
print("\nDictionary to Series:\n", aqi_series)

# 3. Accessing elements (Dictionary-like behavior)
print("\nAQI in HSR Layout:", aqi_series['HSR Layout'])

# 4. Vectorized Math (NumPy-like behavior)
# Adjusting every value simultaneously without writing a 'for' loop
adjusted_aqi = aqi_series + 10
print("\nAdjusted AQI (+10):\n", adjusted_aqi)

# 5. Boolean Filtering
# Selecting only the items that meet a specific condition
high_pollution = aqi_series[aqi_series > 100]
print("\nLocations with AQI > 100:\n", high_pollution)
```

#### **5. What is a DataFrame?**
A **DataFrame** is a two-dimensional, tabular data structure composed of rows and columns, functioning much like a SQL table or an Excel spreadsheet. Under the hood, it is essentially a dictionary of Pandas Series objects that all share the exact same row index.

**Key Characteristics**
* **Heterogeneous Types:** Unlike standard NumPy matrices, each column in a DataFrame can hold a entirely different data type (e.g., a string column next to a float column).
* **Labeled Axes:** Both rows (the index) and columns have explicit labels, allowing for intuitive data alignment, merging, and retrieval.
* **Size-Mutable:** You can dynamically append rows or insert/delete columns without needing to recreate the entire object.

```python
import pandas as pd

# 1. Creating a DataFrame from a dictionary of lists (most common method)
# Each key becomes a column name, and each list becomes the column's data
data = {
    'City': ['Kolkata', 'Bengaluru', 'Chennai'],
    'PM10_Level': [102.4, 89.1, 76.5],
    'Status': ['Moderate', 'Satisfactory', 'Satisfactory']
}
df = pd.DataFrame(data)

print("Standard DataFrame:\n")
display(df)

# 2. Accessing components
# Extracting a single column returns a Series
pm10_series = df['PM10_Level']

# Extracting multiple columns returns a smaller DataFrame
subset_df = df[['City', 'Status']]

# 3. Adding a new column
# You can broadcast a single value or assign a calculated array/list
df['Country'] = 'India'
df['Requires_Action'] = df['PM10_Level'] > 100

print("\nDataFrame after adding columns:\n")
display(df)

# 4. Exploring the underlying structure
print("\nDataFrame Index:", df.index)
print("DataFrame Columns:", df.columns)
print("DataFrame Shape (Rows, Columns):", df.shape)
```

#### **6. What is the difference between a Series and a DataFrame?**
The fundamental difference lies in dimensionality: a Series represents a single column of data (1D), whereas a DataFrame is a multi-column table (2D) composed of multiple aligned Series. Every column you extract from a DataFrame is returned as a Series.

**Key Structural Differences**

| Feature | `pd.Series` | `pd.DataFrame` |
| --- | --- | --- |
| **Dimensions** | 1-Dimensional (Array/List-like) | 2-Dimensional (Table/Grid-like) |
| **Data Types** | Homogeneous (typically a single data type) | Heterogeneous (each column can be a different type) |
| **Conceptual Model** | A single column of data with row labels (index) | A dictionary of Series objects sharing the exact same index |
| **Size Mutability** | Fixed size (adding elements usually requires recreating the object) | Mutable (add or drop columns and rows dynamically) |

```python
import pandas as pd

# 1. Start with a DataFrame (2D)
data = {
    'Station': ['Indiranagar', 'BTM Layout', 'HSR Layout'],
    'PM25_Level': [45.2, 112.5, 89.0]
}
df = pd.DataFrame(data)
print("Original Object Type:", type(df))

# 2. Extracting a single column returns a Series (1D)
pm25_series = df['PM25_Level']
print("\nExtracted Column Type:", type(pm25_series))

# 3. Creating an independent Series
# If you don't specify an index, it defaults to 0, 1, 2... matching the DataFrame
aqi_status = pd.Series(['Good', 'Poor', 'Moderate'], name='Status')
print("\nIndependent Series:\n", aqi_status)

# 4. Combining them: Adding the Series as a new column in the DataFrame
df['AQI_Status'] = aqi_status
print("\nCombined DataFrame:\n")
display(df)
```

#### **7. How do you create a Series?**
You can create a Pandas Series by passing various Python or NumPy data structures into the `pd.Series()` constructor. The constructor accepts the data itself, an optional custom index, and an optional name for the Series.

**Common Data Sources for a Series**
* **Python Lists:** Automatically assigned a numeric index (0, 1, 2...) unless a custom index is provided.
* **Python Dictionaries:** The dictionary keys automatically become the Series index, and the values become the data.
* **NumPy Arrays:** Functions exactly like a list, leveraging NumPy's memory efficiency.
* **Scalar Values:** A single value (like an integer or string) that gets repeated across a provided index.

```python
import pandas as pd
import numpy as np

# 1. Creating from a standard list (default integer index)
kit_sizes = pd.Series(['S', 'S', 'S', 'S'], name='Kit_Size')
print("From List (Default Index):\n", kit_sizes)

# 2. Creating from a list with a custom index
# The length of the index list must exactly match the length of the data list
squad_numbers = pd.Series(
    [10, 11, 10, 11], 
    index=['Messi', 'Raphinha', 'Mac Allister', 'Di Maria'],
    name='Squad_Number'
)
print("\nFrom List (Custom Index):\n", squad_numbers)

# 3. Creating from a Dictionary
# Pandas automatically uses the dictionary keys as the index labels
player_teams = {
    'Messi': 'Argentina',
    'Raphinha': 'FC Barcelona',
    'Mac Allister': 'Liverpool',
    'Di Maria': 'Argentina'
}
team_series = pd.Series(player_teams, name='Team')
print("\nFrom Dictionary:\n", team_series)

# 4. Creating from a NumPy array
# Useful when pulling data generated by mathematical functions
random_data = np.array([4.5, 8.2, 9.1])
np_series = pd.Series(random_data, index=['x', 'y', 'z'])
print("\nFrom NumPy Array:\n", np_series)

# 5. Creating from a Scalar value
# The single value 'Economy' is broadcasted across all provided index labels
flight_class = pd.Series('Economy', index=['Flight_1', 'Flight_2', 'Flight_3'])
print("\nFrom Scalar Value:\n", flight_class)
```

#### **8. How do you create a DataFrame?**
You can construct a DataFrame from scratch using several native Python data structures, though the most common approach is using a dictionary of lists. The `pd.DataFrame()` constructor automatically aligns the provided data into rows and columns.

**Common Data Sources for a DataFrame**
* **Dictionary of Lists:** Column-oriented creation where keys become column headers and lists become the column data.
* **List of Dictionaries:** Row-oriented creation where each dictionary represents a single row (useful when parsing JSON data).
* **2D NumPy Arrays:** Converts mathematical matrices into labeled tables by explicitly providing column and index names.
* **External Files:** In practice, you will most often create DataFrames by reading files using functions like `pd.read_csv()` or `pd.read_excel()`.

```python
import pandas as pd
import numpy as np

# 1. From a Dictionary of Lists (Column-oriented)
# All lists must be the exact same length.
dict_data = {
    'City': ['Bengaluru', 'Kolkata', 'Chennai'],
    'AQI': [85, 102, 76],
    'Status': ['Satisfactory', 'Moderate', 'Satisfactory']
}
df_from_dict = pd.DataFrame(dict_data)
print("From Dictionary of Lists:\n")
display(df_from_dict)

# 2. From a List of Dictionaries (Row-oriented)
# Missing keys in specific dictionaries will automatically become NaN values.
list_data = [
    {'City': 'Bengaluru', 'AQI': 85},
    {'City': 'Kolkata', 'AQI': 102, 'Status': 'Moderate'},
    {'City': 'Chennai', 'AQI': 76, 'Status': 'Satisfactory'}
]
df_from_list = pd.DataFrame(list_data)
print("\nFrom List of Dictionaries:\n")
display(df_from_list)

# 3. From a 2D NumPy Array
# You must explicitly define the column names and index labels, otherwise they default to 0, 1, 2...
numpy_matrix = np.array([
    [85, 45.2], 
    [102, 55.1], 
    [76, 30.5]
])
df_from_numpy = pd.DataFrame(
    numpy_matrix, 
    columns=['AQI', 'PM2.5'], 
    index=['Bengaluru', 'Kolkata', 'Chennai']
)
print("\nFrom 2D NumPy Array:\n")
display(df_from_numpy)
```

#### **9. How do you read data from a CSV file?**
The `pd.read_csv()` function is the workhorse for data ingestion in Pandas. It reads comma-separated values and converts them directly into a DataFrame, offering dozens of parameters to handle dates, specific columns, and missing values on the fly.

*(Note: You would need a local file named `customers.csv` to execute these exact commands).*

```python
import pandas as pd

path = "./../Datasets/"
# 1. Basic Import
# Reads the entire file, assuming the first row contains column headers
df = pd.read_csv(path + 'customers.csv')
display(df.head())

# 2. Reading Specific Columns
# 'usecols' saves memory by only loading the columns you actually need for your analysis
df_subset = pd.read_csv(
    path + 'customers.csv', 
    usecols=['SignupDate', 'Gender', 'Age']
)
display(df_subset.head())

# 3. Parsing Dates on Import
# 'parse_dates' automatically converts standard string dates into Pandas datetime objects
df_dates = pd.read_csv(
    path + 'customers.csv', 
    parse_dates=['SignupDate']
)
display(df_dates.head())
print("Data type of 'date' column:", df_dates['SignupDate'].dtype)

# 4. Setting an Index Column
# 'index_col' assigns a specific column to act as the row labels immediately upon loading
df_indexed = pd.read_csv(
    path + 'customers.csv', 
    index_col='CustomerID'
)
display(df_indexed.head())

# 5. Handling Custom Missing Values
# 'na_values' tells pandas to treat specific strings or numbers (like '-999') as NaN
df_clean = pd.read_csv(
    path + 'customers.csv', 
    na_values=['Missing', 'Not Available', '-999', ' ']
)
display(df_clean.head())
```

#### **10. How do you write a DataFrame to a CSV file?**
The `to_csv()` method exports your DataFrame back into a standard comma-separated text file. It gives you precise control over how row indices, column headers, and missing values are formatted on the disk.

```python
import pandas as pd
import numpy as np

# Assume 'df' is the DataFrame you have been cleaning and processing
df = pd.DataFrame(
    {
        'City': ['Kolkata', 'Noida', 'Bengaluru', 'Hyderabad'],
        'AQI': ['100', '203', '147', '196'],
        'Status': ['Ok', 'Polluted', np.nan, 'Polluted'],
        'Population': [300000, 200000, np.nan, np.nan]
    }
)
# 1. Basic Export
# Writes the entire DataFrame to the current working directory.
# Note: By default, this saves the row index as the very first column in the CSV.
df.to_csv('processed_data.csv')

# 2. Export Without the Index (Most Common)
# Set index=False to prevent Pandas from writing the default numeric index (0, 1, 2...).
# This prevents an unnamed index column from appearing the next time you read the file.
df.to_csv('clean_data_no_index.csv', index=False)

# 3. Export Specific Columns
# Use the 'columns' parameter to save only a subset of your data.
df.to_csv(
    'subset_data.csv', 
    columns=['City', 'AQI', 'Status'], 
    index=False
)

# 4. Customizing Missing Data Representation
# 'na_rep' specifies how NaN values should appear in the text file. 
# By default, they are saved as empty strings. Here we replace them with 'NULL'.
df.to_csv(
    'data_with_nulls.csv', 
    index=False, 
    na_rep='NULL'
)
```

#### **11. How do you display the first few rows of a DataFrame?**
The `head()` method is used to quickly inspect the top rows of a DataFrame. This is an essential step immediately after loading data to verify the schema, column names, and data formats without printing thousands of rows to your screen.

```python
import pandas as pd

# Creating a sample DataFrame with 6 rows
data = {
    'Station': ['Indiranagar', 'BTM Layout', 'HSR Layout', 'Koramangala', 'Whitefield', 'Jayanagar'],
    'AQI': [85, 112, 142, 110, 155, 92],
    'Status': ['Satisfactory', 'Poor', 'Very Poor', 'Poor', 'Very Poor', 'Satisfactory']
}
df = pd.DataFrame(data)

# 1. Default Usage
# By default, head() returns exactly the first 5 rows of the DataFrame
print("Default (First 5 rows):\n")
display(df.head())

# 2. Custom Row Count
# Pass an integer 'n' to view exactly that many rows from the top
print("\nFirst 3 rows:\n")
display(df.head(3))

# 3. Inspecting the Bottom Rows
# The tail() method works exactly like head(), but returns rows from the end of the dataset
# This is useful for checking if data appended correctly to the bottom of your file
print("\nLast 2 rows:\n")
display(df.tail(2))
```

#### **12. How do you display the last few rows?**
The `tail()` method is the exact counterpart to `head()`, used to inspect the bottom rows of a DataFrame or Series. It is particularly useful for verifying the most recently appended data in chronological datasets or checking if summary/total rows were accidentally imported at the end of a CSV file.

```python
import pandas as pd

# Creating a sample time-series DataFrame
data = {
    'Date': ['2026-09-07', '2026-09-08', '2026-09-09', '2026-09-10', '2026-09-11', '2026-09-12'],
    'Station': ['Kolkata Central'] * 6,
    'PM25_Level': [55.2, 58.1, 62.4, 70.1, 75.8, 80.2]
}
df = pd.DataFrame(data)

# 1. Default Usage
# By default, tail() returns exactly the last 5 rows of the DataFrame
print("Default (Last 5 rows):\n")
display(df.tail())

# 2. Custom Row Count
# Pass an integer 'n' to view exactly that many rows from the very bottom
print("\nLast 2 rows (Most recent data):\n")
display(df.tail(2))

# 3. Using tail() on a Single Column (Series)
# The method works identically on a 1D Series
print("\nLast 3 PM2.5 readings:\n")
print(df['PM25_Level'].tail(3))
```

#### **13. What does df.info() do?**
The `info()` method provides a concise structural summary of a DataFrame. It is the best diagnostic tool to run immediately after loading a dataset to check for missing values and ensure data types were parsed correctly.

**Key Diagnostic Information Provided**

* **Dimensions:** The total number of rows (index entries) and columns.
* **Data Types:** The `dtype` of each column (e.g., `int64`, `float64`, `object` for strings).
* **Non-Null Counts:** How many valid (non-missing) entries exist in each column, which instantly highlights where data is missing.
* **Memory Usage:** The RAM footprint of the DataFrame.

```python
import pandas as pd
import numpy as np

# Creating a dataset with mixed types and missing values (NaN)
data = {
    'station_id': [101, 102, 103, 104],
    'city': ['Bengaluru', 'Kolkata', 'Chennai', 'Mumbai'],
    'pm25_level': [45.2, 55.1, np.nan, 80.2],  # Note the missing value here
    'is_active': [True, True, False, True]
}
df = pd.DataFrame(data)

# 1. Standard Info Summary
# Notice how 'pm25_level' will show 3 non-null out of 4 total rows
print("Standard df.info() Output:\n")
df.info()

# 2. Deep Memory Profiling
# By default, Pandas only estimates the memory of 'object' (string) columns. 
# Using memory_usage='deep' forces Pandas to inspect the exact memory footprint of all strings.
print("\nDeep Memory Profiling:\n")
df.info(memory_usage='deep')
```

#### **14. What does df.describe() return?**
The `describe()` method generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution, automatically excluding `NaN` values. By default, it only profiles numeric columns (integers and floats), providing a rapid mathematical overview of your data.

**Numeric Output Metrics**
* **count:** Number of non-null observations.
* **mean:** The average value.
* **std:** Standard deviation (measure of data spread).
* **min / max:** The lowest and highest values.
* **25%, 50%, 75%:** The lower percentile, median (50%), and upper percentile.

```python
import pandas as pd
import numpy as np

# Creating a dataset with both numeric and categorical data
data = {
    'station': ['Indiranagar', 'BTM Layout', 'HSR Layout', 'Koramangala', 'Whitefield'],
    'aqi': [85, 112, 142, 110, 155],
    'pm25_level': [45.2, 55.1, 70.4, np.nan, 82.5],
    'status': ['Satisfactory', 'Poor', 'Very Poor', 'Poor', 'Very Poor']
}
df = pd.DataFrame(data)

# 1. Default Behavior (Numeric Columns Only)
print("Numeric Statistics:\n")
display(df.describe())

# 2. Profiling Categorical/String Data
# Passing include='object' forces describe() to analyze string columns.
# It returns count, number of unique values, the most frequent value (top), 
# and its frequency (freq).
print("\nCategorical Statistics:\n")
display(df.describe(include=['object']))

# 3. Profiling All Columns
# Passing include='all' analyzes every column, filling cells with NaN 
# where a metric doesn't apply (e.g., 'mean' for a string column).
print("\nAll Columns Summary:\n")
display(df.describe(include='all'))
```

#### **15. How do you check the shape of a DataFrame?**
The `shape` attribute returns a tuple representing the dimensionality of the DataFrame in the format `(rows, columns)`. Because it is an underlying property of the DataFrame rather than a functional method, you do not use parentheses when calling it.

**Key Characteristics**
* **Zero-Indexed Tuple:** The first element `[0]` represents the total number of rows (index length), and the second element `[1]` represents the total number of columns.
* **Series Shape:** If called on a 1D Series, it returns a tuple with a single element: `(rows,)`.

```python
import pandas as pd

# Creating a dataset with 4 rows and 3 columns
data = {
    'Station': ['BTM Layout', 'HSR Layout', 'Koramangala', 'Indiranagar'],
    'PM10_Level': [102.4, 89.1, 76.5, 95.0],
    'Status': ['Moderate', 'Satisfactory', 'Satisfactory', 'Moderate']
}
df = pd.DataFrame(data)

# 1. Getting the full shape tuple
dimensions = df.shape
print(f"Full Shape Tuple: {dimensions}")

# 2. Extracting rows and columns programmatically
# Useful when setting up loops or validating data sizes before machine learning pipelines
num_rows = df.shape[0]
num_cols = df.shape[1]

print(f"Total Rows: {num_rows}")
print(f"Total Columns: {num_cols}")

# 3. Checking the shape of a single column (Series)
series_shape = df['PM10_Level'].shape
print(f"Series Shape: {series_shape}")
```

#### **16. How do you list column names?**
The `columns` attribute returns the column labels of a DataFrame. Like `shape`, it is an attribute rather than a method, so you do not use parentheses. By default, it returns a pandas `Index` object, but it can easily be converted into a standard Python list.

**Key Characteristics**
* **Index Object:** The returned object is optimized for pandas operations but behaves similarly to a list.
* **Iterable:** You can easily loop through the `columns` attribute to apply bulk text formatting (like converting all names to lowercase).

```python
import pandas as pd

# Creating a sample dataset
data = {
    'Station_ID': ['BTM', 'HSR', 'KOR'],
    'PM25_Level': [45.2, 55.1, 70.4],
    'Air_Quality_Index': [85, 112, 142]
}
df = pd.DataFrame(data)

# 1. Accessing the columns attribute directly
print("Raw Columns Index Object:")
print(df.columns)

# 2. Converting to a standard Python list (Most Common)
# Using .tolist() is the preferred way to extract headers for iteration or manipulation
col_list = df.columns.tolist()
print("\nAs a Python List:")
print(col_list)

# 3. Iterating through column names
# Useful for bulk renaming, such as converting all headers to lowercase
print("\nIterating through columns:")
for col in df.columns:
    print(f" - {col.lower()}")

# 4. Modifying column names directly
# You can overwrite the entire columns attribute by assigning a list of the exact same length
df.columns = ['station', 'pm25', 'aqi']
print("\nRenamed Columns:")
print(df.columns.tolist())
```

#### **17. How do you check data types of columns?**
The `dtypes` attribute returns a Series displaying the specific data type of each column in your DataFrame. Understanding these types is critical, as mathematical functions will fail on string columns, and date functions will fail if a date is stored as an `object` (string) instead of a `datetime` object.

**Common Pandas Data Types**
* **`int64` / `float64`:** Integer and floating-point numbers.
* **`object`:** Typically represents strings, but can also indicate a column containing mixed data types.
* **`bool`:** True/False boolean values.
* **`datetime64`:** Date and time values, which unlock time-series methods.

```python
import pandas as pd

# Creating a dataset with a variety of data types
data = {
    'station_name': ['Bally', 'New Town', 'Salt Lake'], # object (string)
    'aqi': [110, 85, 92],                               # int64
    'pm25_level': [45.2, 38.1, 40.5],                   # float64
    'is_active': [True, True, False],                   # bool
    'last_calibrated': ['2026-08-01', '2026-08-15', '2026-09-01'] # object (currently a string)
}
df = pd.DataFrame(data)

# Convert the string column to actual datetime objects
df['last_calibrated'] = pd.to_datetime(df['last_calibrated'])

# 1. Checking all data types
# Returns a Pandas Series where the index is column names and values are the data types
print("Data types for all columns:\n")
print(df.dtypes)

# 2. Checking a single column's data type
print(f"\nData type of 'pm25_level': {df['pm25_level'].dtype}")

# 3. Filtering columns by data type
# select_dtypes() allows you to isolate columns for specific operations 
# (e.g., extracting only numeric columns before applying a scaling algorithm)
numeric_df = df.select_dtypes(include=['int64', 'float64'])
print("\nNumeric columns only:\n")
display(numeric_df)

# You can also use 'number' as a shortcut for all int/float types, 
# or use 'exclude' to drop certain types
text_only = df.select_dtypes(exclude=['number', 'bool', 'datetime64'])
print("\nText (object) columns only:\n")
display(text_only)
```

#### **18. How do you rename columns?**
You can rename columns in Pandas by either mapping specific column names using a dictionary or by completely overwriting the column index array.

**Common Renaming Methods**
* **The `rename()` Method:** The safest approach. It uses a dictionary to target specific columns, leaving unspecified columns untouched regardless of their order.
* **Index Assignment:** Overwrites the entire `df.columns` attribute at once. It is fast but requires providing a list that exactly matches the total number of columns.
* **Vectorized String Operations:** Applies string methods (like lowercase or replace) across the entire column index simultaneously, which is ideal for cleaning messy, inconsistent headers.

```python
import pandas as pd

# Creating a sample dataset with messy, capitalized column names containing spaces
data = {
    'Station Name': ['Salt Lake', 'New Town'],
    'PM2.5': [45.2, 38.1],
    'AQI Level': [92, 85]
}
df = pd.DataFrame(data)
print("Original Columns:", df.columns.tolist())

# 1. Using rename() with a Dictionary (Targeted approach)
# Keys are the old names, values are the new names.
# This returns a new DataFrame unless you pass inplace=True.
df_renamed = df.rename(columns={
    'Station Name': 'station', 
    'PM2.5': 'pm25'
})
print("\nTargeted Rename:")
display(df_renamed)

# 2. Overwriting the columns attribute (Complete replacement)
# The provided list must match the exact number of columns in the DataFrame.
df.columns = ['station_id', 'pm25_reading', 'aqi_score']
print("\nFull Overwrite:")
display(df)

# 3. Bulk Vectorized String Operations (Automated cleanup)
# Ideal for standardizing hundreds of columns without typing them out manually.
df_raw = pd.DataFrame(data)

# Convert all headers to lowercase, replace spaces with underscores, and remove periods
df_raw.columns = (
    df_raw.columns
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('.', '')
)

print("\nAutomated String Cleanup:")
display(df_raw)
```

#### **21. What is the difference between loc[] and iloc[]?**
The fundamental difference between `loc[]` and `iloc[]` is how they locate data: **`loc[]` uses explicit labels (names)**, while **`iloc[]` uses implicit integer positions (0-based indexing)**.

Both methods accept two arguments in the format `[rows, columns]`, but they process those arguments very differently.

**Key Behavioral Differences**  
| Feature | `loc[]` (Label-Based) | `iloc[]` (Integer-Based) |
| --- | --- | --- |
| **Lookup Mechanism** | Looks for the actual name/label of the row index and column header. | Looks for the exact numerical row and column position (0, 1, 2...). |
| **Slicing Bounds** | Both the start and stop bounds are **INCLUSIVE**. | The stop bound is **EXCLUSIVE** (matches standard Python list behavior). |
| **Valid Inputs** | Strings, lists of strings, boolean arrays, or a slice object with labels (`'A':'C'`). | Integers, lists of integers, boolean arrays, or a slice object with integers (`0:3`). |

```python
import pandas as pd

# Creating a dataset with custom row index labels (not standard 0,1,2..)
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally', 'Park Street'],
    'aqi': [115, 85, 92, 110, 105],
    'pm25': [48.2, 38.1, 40.5, 45.0, 42.1]
}
# Setting letters as the row index to clearly distinguish label from position
df = pd.DataFrame(data, index=['a', 'b', 'c', 'd', 'e'])
display(df)

print("\n1. Single Row Selection")
# loc looks for the label 'b'
print("loc['b']:\n", df.loc['b'].to_frame().T)
# iloc looks for the integer position 1 (which happens to be row 'b')
print("\niloc[1]:\n", df.iloc[1].to_frame().T)

print("\n2. Selecting Rows and Columns")
# loc uses the column name string
print("loc['c', 'aqi']:", df.loc['c', 'aqi'])
# iloc uses the column integer position (aqi is at index 1)
print("iloc[2, 1]:", df.iloc[2, 1])

print("\n3. Slicing Behavior (Crucial Difference)")
# loc includes the final element 'c'
print("loc['a':'c'] (INCLUSIVE of 'c'):\n")
display(df.loc['a':'c', ['station', 'aqi']])

# iloc excludes the final index 2 (stops at 1)
print("\niloc[0:2] (EXCLUSIVE of index 2):\n")
display(df.iloc[0:2, 0:2])
```

#### **22. What is label-based indexing?**
Label-based indexing is a method of extracting data from a DataFrame or Series using the explicit names (labels) assigned to rows and columns, rather than their underlying numerical positions. In Pandas, this operation is executed exclusively using the `.loc[]` indexer.

**Key Characteristics of `.loc[]`**  
* **Explicit Naming:** It strictly searches for the actual row index label (which could be a string, a date, or an integer) and the exact column header string.
* **Inclusive Slicing:** When slicing data (e.g., `loc['A':'C']`), Pandas includes both the start bound and the stop bound. This is a major departure from standard Python list slicing.
* **Boolean Masking:** It natively accepts boolean arrays, making it the primary method for conditional data filtering.

```python
import pandas as pd

# Creating a dataset and setting explicit string labels as the row index
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally'],
    'aqi': [115, 85, 92, 110],
    'pm25_level': [48.2, 38.1, 40.5, 45.0]
}
# Assigning custom row labels instead of the default 0, 1, 2, 3
df = pd.DataFrame(data, index=['Day_1', 'Day_2', 'Day_3', 'Day_4'])
print("Original Labeled DataFrame:")
display(df)

# 1. Selecting a single row by its label
# Returns a Series where the column names become the index
day2_data = df.loc['Day_2']
print("\nSingle Row (Day_2):\n", day2_data)

# 2. Selecting specific rows and specific columns
# Syntax: df.loc[[row_labels], [column_labels]]
subset = df.loc[['Day_1', 'Day_3'], ['station', 'aqi']]
print("\nSpecific Rows and Columns:")
display(subset)

# 3. Slicing with labels (INCLUSIVE of the end bound)
# Extracts everything from Day_2 through Day_4, including Day_4
slice_df = df.loc['Day_2':'Day_4', 'aqi':'pm25_level']
print("\nInclusive Slicing:")
display(slice_df)

# 4. Conditional Filtering (Boolean Indexing)
# Extract all columns where the AQI is greater than 100
high_pollution = df.loc[df['aqi'] > 100, ['station', 'aqi']]
print("\nConditional Filtering with loc:")
display(high_pollution)
```

#### **23. What is integer-based indexing?**
Integer-based indexing is a method of extracting data based solely on its 0-based numerical position in the DataFrame, completely ignoring any custom row labels or column names. In Pandas, this operation is executed exclusively using the `.iloc[]` indexer.

**Key Characteristics of `.iloc[]`**  
* **Positional Targeting:** It operates exactly like standard Python lists or NumPy arrays. The first row and first column are always at index `0`.
* **Exclusive Slicing:** When slicing data (e.g., `iloc[0:3]`), Pandas extracts rows 0, 1, and 2, but strictly excludes the stop bound (index 3).
* **Negative Indexing:** It natively supports negative integers, allowing you to easily grab the last row (`iloc[-1]`) or last column without needing to know the exact length of the dataset.

```python
import pandas as pd

# Creating a dataset and setting explicit string labels
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally'],
    'aqi': [115, 85, 92, 110],
    'pm25_level': [48.2, 38.1, 40.5, 45.0]
}
df = pd.DataFrame(data, index=['Day_1', 'Day_2', 'Day_3', 'Day_4'])
print("Original Labeled DataFrame:")
display(df)

# 1. Selecting a single row by position
# Extracts the second row (index 1), ignoring the label 'Day_2'
row_one = df.iloc[1]
print("\nRow at index 1:\n", row_one)

# 2. Selecting specific rows and columns by position
# Syntax: df.iloc[[row_indices], [column_indices]]
# Extracts the 1st and 3rd rows, and the 1st and 2nd columns
subset = df.iloc[[0, 2], [0, 1]]
print("\nSpecific Row/Col Positions:")
display(subset)

# 3. Slicing with integers (EXCLUSIVE of the end bound)
# Extracts rows 1 and 2, and columns 1 and 2
slice_df = df.iloc[1:3, 1:3]
print("\nExclusive Slicing (Rows 1:3, Cols 1:3):")
display(slice_df)

# 4. Negative Indexing
# Extracts the very last row and the last two columns
last_row_data = df.iloc[-1, -2:]
print("\nNegative Indexing (Last Row, Last 2 Cols):\n", last_row_data)
```

#### **24. How do you filter rows based on conditions?**
Filtering rows relies on a technique called **boolean masking**. You create a condition that evaluates every row as `True` or `False`, and Pandas returns a new DataFrame containing only the rows where the condition evaluated to `True`.

**Key Filtering Rules**  
* **Bitwise Operators:** You cannot use standard Python logical operators (`and`, `or`, `not`). You must use bitwise operators: `&` (AND), `|` (OR), and `~` (NOT).
* **Mandatory Parentheses:** When chaining multiple conditions, wrap each individual condition in parentheses `()` to dictate the order of operations, otherwise Pandas will throw an error.
* **The `.isin()` Method:** A cleaner alternative to writing multiple `|` (OR) statements when checking if a column value matches any item in a list.

```python
import pandas as pd

# Creating a sample dataset
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally', 'Park Street'],
    'aqi': [115, 85, 92, 160, 105],
    'status': ['Poor', 'Good', 'Satisfactory', 'Very Poor', 'Poor']
}
df = pd.DataFrame(data)

# 1. Single Condition Filtering
# Extract all rows where the AQI is strictly greater than 100
high_aqi = df[df['aqi'] > 100]
print("Single Condition (AQI > 100):")
display(high_aqi)

# 2. Multiple Conditions using AND (&)
# Both conditions must be True. Notice the mandatory parentheses.
moderate_risk = df[(df['aqi'] > 90) & (df['aqi'] < 120)]
print("\nMultiple Conditions AND:")
display(moderate_risk)

# 3. Multiple Conditions using OR (|)
# Only one condition needs to be True.
extreme_cases = df[(df['aqi'] < 90) | (df['status'] == 'Very Poor')]
print("\nMultiple Conditions OR:")
display(extreme_cases)

# 4. Using the .isin() Method
# Checks if the column value exists within a provided list. 
# Much cleaner than writing: (df['station'] == 'Howrah') | (df['station'] == 'Bally')
target_stations = ['Howrah', 'Bally']
specific_locations = df[df['station'].isin(target_stations)]
print("\nUsing .isin() for Lists:")
display(specific_locations)

# 5. Using String Methods for Filtering
# Extracts rows where the status text contains the word 'Poor'
poor_conditions = df[df['status'].str.contains('Poor')]
print("\nString Method Filtering:")
display(poor_conditions)
```

#### **25. How do you select rows with multiple conditions?**
Selecting rows based on multiple conditions requires combining individual boolean masks using bitwise operators.

Standard Python logical operators (`and`, `or`, `not`) will throw a `ValueError` because they attempt to evaluate the entire DataFrame's truth value at once. You must use element-wise operators instead, and **you must wrap every individual condition in parentheses `()`** due to Python's operator precedence rules.

**Key Operators for Multiple Conditions**  
* **`&` (AND):** The row is selected only if *all* conditions are True.
* **`|` (OR):** The row is selected if *at least one* condition is True.
* **`~` (NOT):** Reverses the condition (True becomes False).
* **`.query()`:** An elegant, string-based alternative method that often makes complex multi-condition filters much easier to read.

```python
import pandas as pd

# Creating a dataset with multiple numerical and categorical columns
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally', 'Park Street'],
    'aqi': [115, 85, 92, 160, 105],
    'pm25': [48.2, 35.1, 39.5, 65.0, 42.1],
    'is_active': [True, True, False, True, True]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Multiple Conditions with AND (&)
# Both conditions must be True. (Notice the parentheses!)
high_and_active = df[(df['aqi'] > 100) & (df['is_active'] == True)]
print("\nStations with AQI > 100 AND are Active:")
display(high_and_active)

# 2. Multiple Conditions with OR (|)
# Only one condition needs to be True.
extreme_or_inactive = df[(df['aqi'] >= 150) | (df['is_active'] == False)]
print("\nStations with AQI >= 150 OR are Inactive:")
display(extreme_or_inactive)

# 3. Combining AND, OR, and NOT (~)
# Select stations that are active, AND (AQI is < 90 OR PM2.5 is < 40)
complex_filter = df[~df['is_active'] == False & ((df['aqi'] < 90) | (df['pm25'] < 40))]
print("\nComplex Filter Using Multiple Operators:")
display(complex_filter)

# 4. The .query() Method (Cleaner Syntax Alternative)
# You can write the entire logic as a string. It allows the use of 'and' / 'or' keywords.
# Note: It requires variable names to be valid Python identifiers (no spaces).
query_result = df.query("aqi > 100 and is_active == True")
print("\nUsing .query() for cleaner syntax:")
display(query_result)
```

#### **26. What is boolean indexing?**
Boolean indexing is the process of selecting subsets of data based on their actual values rather than their row labels or integer positions. It relies on a "boolean mask"—a sequence of `True` and `False` values that perfectly aligns with the rows of your dataset.

When you pass this mask into a DataFrame or Series, Pandas acts like a sieve: it keeps every row corresponding to `True` and drops every row corresponding to `False`.

**Key Characteristics**  
* **The Mask:** A boolean mask is simply a Pandas Series containing only boolean data types (`bool`).
* **Dynamic Generation:** In practice, you rarely type out `True` and `False` manually. Instead, you generate the mask dynamically by applying a logical condition to a column (e.g., `df['AQI'] > 100`).
* **Alignment:** The mask must have the exact same number of rows as the DataFrame you are filtering, or Pandas will throw an error.

```python
import pandas as pd

# Creating a sample dataset
data = {
    'station': ['Howrah', 'New Town', 'Salt Lake', 'Bally'],
    'aqi': [115, 85, 92, 160]
}
df = pd.DataFrame(data)
print("Original Dataset:\n")
display(df)

# 1. The Manual Boolean Mask
# Passing an explicit list of True/False values of the exact same length as the DataFrame
manual_mask = [True, False, False, True]
explicit_filter = df[manual_mask]

print("\nFiltering with a Manual Mask (Keeps rows 0 and 3):")
display(explicit_filter)

# 2. The Dynamic Boolean Mask
# This is what happens under the hood when you write a condition.
# It evaluates the column and returns a Series of True/False values.
dynamic_mask = df['aqi'] > 100
print("\nThe Generated Boolean Series:")
print(dynamic_mask)
print(f"\nData Type of the mask: {dynamic_mask.dtype}")

# 3. Applying the Dynamic Mask
# We pass the generated Series into the DataFrame brackets
# This is the exact equivalent of: df[df['aqi'] > 100]
applied_filter = df[dynamic_mask]

print("\nFiltering with the Dynamic Mask:")
display(applied_filter)
```

#### **27. How do you sort a DataFrame?**
You can sort a DataFrame in Pandas using two primary methods: `sort_values()` to sort by the actual data within columns, and `sort_index()` to sort by the row labels.

By default, Pandas returns a completely new, sorted DataFrame and leaves your original dataset unchanged, unless you explicitly pass the parameter `inplace=True`.

**Key Sorting Parameters**  
* **`by`:** The column name (or list of column names) to sort by.
* **`ascending`:** A boolean (or list of booleans) dictating the sort order. Defaults to `True` (A-Z, 0-9).
* **`na_position`:** Dictates where missing values (`NaN`) should end up. Defaults to `'last'`, but can be changed to `'first'`.

```python
import pandas as pd
import numpy as np

# Creating a dataset with a tie in AQI and a missing value
data = {
    'station': ['Salt Lake', 'New Town', 'Bally', 'Howrah', 'Park Street'],
    'aqi': [92, 85, 110, 110, np.nan],
    'pm25_level': [40.5, 38.1, 45.0, 48.2, 42.1]
}
# Using a custom index out of alphabetical order
df = pd.DataFrame(data, index=['C', 'A', 'E', 'B', 'D'])
print("Original Dataset:")
display(df)

# 1. Sorting by a single column (Default: Ascending)
# Notice how the NaN value is automatically pushed to the very bottom
sort_single = df.sort_values(by='aqi')
print("\nSorted by AQI (Ascending):")
display(sort_single)

# 2. Sorting in Descending Order
sort_desc = df.sort_values(by='aqi', ascending=False)
print("\nSorted by AQI (Descending):")
display(sort_desc)

# 3. Sorting by Multiple Columns (Tie-Breaking)
# If the AQIs are tied (like Bally and Howrah), it looks at the pm25_level.
# We can pass a list of booleans to sort AQI ascending, but PM2.5 descending.
sort_multi = df.sort_values(
    by=['aqi', 'pm25_level'], 
    ascending=[True, False]
)
print("\nSorted by Multiple Columns (AQI Asc, PM2.5 Desc):")
display(sort_multi)

# 4. Controlling Missing Values
# Force NaN values to appear at the top of the dataset instead of the bottom
sort_nan = df.sort_values(by='aqi', na_position='first')
print("\nSorted by AQI with NaNs First:")
display(sort_nan)

# 5. Sorting by the Index
# Reorders the DataFrame based on the row labels ('A', 'B', 'C'...) rather than the data
sort_idx = df.sort_index()
print("\nSorted Alphabetically by Row Index:")
display(sort_idx)
```

| Feature | `sort_values()` | `sort_index()` |

#### **29. How do you remove duplicate rows?**
The `drop_duplicates()` method scans your DataFrame for identical rows and removes them. By default, it evaluates every single column to find exact row-to-row matches and keeps only the first occurrence it encounters.

**Key Parameters for Precision Control**  
* **`subset`:** A column name or list of columns. Instead of requiring the entire row to be identical, Pandas will only look for duplicates within these specific columns.
* **`keep`:** Dictates which version of the duplicate to retain.
    * `'first'` (Default): Keeps the initial occurrence.
    * `'last'`: Keeps the final occurrence (useful if the last row is the most recent update).
    * `False`: Drops *all* rows that have duplicates, keeping only truly unique records.
* **`inplace`:** Set to `True` to overwrite your existing DataFrame instead of creating a new one.

```python
import pandas as pd

# Creating a dataset with exact duplicates and partial duplicates
data = {
    'station': ['Salt Lake', 'New Town', 'Salt Lake', 'Bally', 'New Town'],
    'date': ['2026-09-10', '2026-09-10', '2026-09-10', '2026-09-11', '2026-09-10'],
    'aqi': [92, 85, 92, 110, 88] 
    # Row 0 and Row 2 are EXACT duplicates.
    # Row 1 and Row 4 share the same station and date, but have different AQIs.
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Default Behavior (Exact row matches only)
# Removes Row 2 because it is completely identical to Row 0.
exact_deduped = df.drop_duplicates()
print("\nDefault Drop (Keeps first exact match):")
display(exact_deduped)

# 2. Using 'subset' to define what constitutes a duplicate
# Even though the AQI is different, we consider it a duplicate if the 
# station and date are the same.
subset_deduped = df.drop_duplicates(subset=['station', 'date'])
print("\nSubset Drop (Checks only station and date):")
display(subset_deduped)

# 3. Changing the 'keep' parameter
# Drops the first occurrence of the New Town duplicate and keeps the last one (Row 4)
keep_last = df.drop_duplicates(subset=['station', 'date'], keep='last')
print("\nKeep Last Occurrence:")
display(keep_last)

# 4. Dropping ALL duplicates
# Removes any record that appears more than once based on the subset.
# Salt Lake and New Town are completely removed. Only Bally remains.
drop_all = df.drop_duplicates(subset=['station', 'date'], keep=False)
print("\nDrop All Duplicates (Keep only unique records):")
display(drop_all)
```

#### **30. How do you count duplicates?**
You can count the total number of duplicate rows in a DataFrame by chaining the `duplicated()` method with the `sum()` function.

The `duplicated()` method returns a boolean mask—a Series of `True` and `False` values indicating whether each row is a duplicate. Because Python evaluates `True` as `1` and `False` as `0`, summing this boolean mask yields the exact integer count of duplicate records.

**Common Duplicate Counting Techniques**  
* **Total Count (`df.duplicated().sum()`):** Returns a single integer representing how many exact row matches exist.
* **Subset Count:** Passing the `subset` parameter counts duplicates based only on specific columns (e.g., finding multiple readings for the same station on the same day).
* **Frequency Breakdown (`value_counts()`):** While `duplicated().sum()` gives you a single grand total, `value_counts()` allows you to see exactly *which* rows are duplicated and how many times each one appears.

```python
import pandas as pd

# Creating a dataset with exact duplicates and partial duplicates
data = {
    'station': ['Salt Lake', 'New Town', 'Salt Lake', 'Bally', 'New Town', 'New Town'],
    'date': ['2026-09-10', '2026-09-10', '2026-09-10', '2026-09-11', '2026-09-10', '2026-09-10'],
    'aqi': [92, 85, 92, 110, 88, 85] 
}
df = pd.DataFrame(data)

# 1. Count Total Exact Duplicates
# Evaluates all columns. Returns 2 because Salt Lake (Row 2) and New Town (Row 5) 
# are exact copies of Row 0 and Row 1.
total_dupes = df.duplicated().sum()
print(f"Total exact duplicate rows: {total_dupes}")

# 2. Count Duplicates Based on Specific Columns
# Evaluates only 'station' and 'date'. Returns 3 because New Town on 2026-09-10 
# appears three times (leaving 2 duplicates to be counted).
subset_dupes = df.duplicated(subset=['station', 'date']).sum()
print(f"Total duplicate records for the same station/date: {subset_dupes}")

# 3. View the Actual Boolean Mask
# Useful for filtering the DataFrame to look at the duplicates directly
is_duplicate = df.duplicated(subset=['station', 'date'], keep=False)
print("\nRows flagged as duplicates (including originals):")
display(df[is_duplicate].sort_values(by='station'))

# 4. Detailed Frequency Breakdown
# Use value_counts() to see exactly how many times each station/date combination appears.
# The reset_index(name='count') turns the Series back into a clean DataFrame.
duplicate_breakdown = df[['station', 'date']].value_counts().reset_index(name='count')
print("\nFrequency of each record combination:")
display(duplicate_breakdown)
```

#### **31. How do you handle missing values?**
Handling missing values (`NaN`) in Pandas fundamentally relies on three core operations: detecting the gaps, dropping incomplete records, or imputing (filling) the missing data with calculated replacements.

Because air quality datasets frequently suffer from sensor downtime, knowing how to handle gaps without aggressively deleting valuable surrounding data is critical.

**Core Missing Data Methods**  
* **`isna()` / `isnull()`:** Returns a boolean mask of your DataFrame where `True` represents a missing value.
* **`dropna()`:** Deletes entire rows (or columns) that contain missing values.
* **`fillna()`:** Replaces `NaN` with a specific static value, a statistical metric (like the mean), or a directional fill (copying the previous row's value).
* **`interpolate()`:** Estimates missing values by drawing a mathematical line between the data points before and after the gap—ideal for time-series sensor data.

```python
import pandas as pd
import numpy as np

# Creating a time-series dataset with missing sensor readings
data = {
    'date': pd.date_range(start='2026-09-07', periods=6),
    'station': ['Koramangala'] * 6,
    'aqi': [110, np.nan, 115, 120, np.nan, 130],
    'pm25': [45.0, 48.2, np.nan, np.nan, 55.1, 60.0]
}
df = pd.DataFrame(data)
print("Original Dataset with Missing Values:")
display(df)

# 1. Detecting Missing Values
# Chaining isna() with sum() gives a total count of missing values per column
missing_counts = df.isna().sum()
print("\nMissing Values per Column:\n", missing_counts)

# 2. Dropping Incomplete Rows
# The 'subset' parameter ensures we only drop the row if 'aqi' is missing. 
# If 'pm25' is missing but 'aqi' is intact, the row is kept.
dropped_df = df.dropna(subset=['aqi'])
print("\nDropped rows missing AQI data:")
display(dropped_df)

# 3. Statistical Imputation (Mean/Median)
# Replacing missing PM2.5 values with the mathematical average of the column
mean_pm25 = df['pm25'].mean()
df_mean_fill = df.copy()
df_mean_fill['pm25'] = df_mean_fill['pm25'].fillna(mean_pm25)
print(f"\nFilled PM2.5 with Column Mean ({mean_pm25:.2f}):")
display(df_mean_fill[['date', 'pm25']])

# 4. Forward Filling (ffill)
# Replaces NaN with the last known valid observation. 
# Highly common in sensor data when a sensor temporarily goes offline.
df_ffill = df.copy()
df_ffill['aqi'] = df_ffill['aqi'].ffill()
print("\nForward Filled AQI (Carries previous value down):")
display(df_ffill[['date', 'aqi']])

# 5. Linear Interpolation
# Draws a straight line between the points surrounding the NaN.
# If Day 1 is 48.2 and Day 4 is 55.1, it evenly calculates the missing Days 2 and 3.
df_interp = df.copy()
df_interp['pm25'] = df_interp['pm25'].interpolate()
print("\nInterpolated PM2.5 Data:")
display(df_interp[['date', 'pm25']])
```

#### **32. Difference between dropna() and fillna()?**
The fundamental difference lies in their approach to data integrity: `dropna()` **removes** incomplete records entirely, whereas `fillna()` **retains** the records by imputing (guessing or assigning) substitute values for the missing data.

| Feature | `dropna()` (Deletion) | `fillna()` (Imputation) |
| --- | --- | --- |
| **Action** | Deletes rows (or columns) containing `NaN`. | Replaces `NaN` with a specified value or calculation. |
| **Dataset Size** | Reduces the total number of rows/columns. | Keeps the exact original dimensions of the dataset. |
| **Statistical Risk** | Loss of valuable data in the non-missing columns of the dropped row. | Introduction of bias or artificial trends if the substituted value is inaccurate. |
| **Best Use Case** | When missing data represents a tiny fraction of a massive dataset, or when a row is completely blank. | When data is scarce, time-series continuity is required, or the missing value can be logically deduced (e.g., filling missing sensor data with the column mean). |

```python
import pandas as pd
import numpy as np

# Creating a dataset where one sensor failed temporarily
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town'],
    'aqi': [115, np.nan, 92, 85],
    'pm25': [48.2, np.nan, 40.5, 38.1]
}
df = pd.DataFrame(data)
print(f"Original Shape: {df.shape}")
display(df)

# 1. The dropna() Approach
# The entire row for 'Bally' is completely removed from the dataset.
# The DataFrame shrinks from 4 rows to 3 rows.
dropped_df = df.dropna()
print(f"\nAfter dropna() Shape: {dropped_df.shape}")
display(dropped_df)

# 2. The fillna() Approach
# The row for 'Bally' is preserved. The missing numeric values are 
# replaced with a static placeholder (e.g., 0) or a statistical metric.
# The DataFrame remains at 4 rows.
filled_df = df.fillna(0)
print(f"\nAfter fillna(0) Shape: {filled_df.shape}")
display(filled_df)
```

#### **33. How do you replace missing values with the mean?**
Replacing missing values with the mean requires calculating the mathematical average of a column using `.mean()` and passing that result directly into `.fillna()`.

It is highly recommended to perform this operation on a specific column or via a dictionary mapping, rather than applying it across the entire DataFrame. Attempting to run `df.fillna(df.mean())` globally will throw a `TypeError` in modern Pandas if your dataset contains string or categorical columns.

```python
import pandas as pd
import numpy as np

# Creating a dataset with missing sensor readings
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town', 'Park Street'],
    'aqi': [115, np.nan, 92, np.nan, 105],
    'pm25': [48.2, 45.0, np.nan, 38.1, 42.1]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Single Column Imputation (Recommended)
# Calculate the mean of the column (Pandas automatically ignores NaNs during calculation)
aqi_mean = df['aqi'].mean()
print(f"\nCalculated AQI Mean: {aqi_mean}")

# Fill the missing values and reassign back to the column
df['aqi'] = df['aqi'].fillna(aqi_mean)

print("\nAfter Filling AQI with Mean:")
display(df)

# 2. Multiple Column Imputation Using a Dictionary
# This is the cleanest way to fill multiple columns with their respective means simultaneously
# Let's reset the dataframe for this example
df_raw = pd.DataFrame(data)

# Create a dictionary where the key is the column name and the value is the calculated mean
fill_values = {
    'aqi': df_raw['aqi'].mean(),
    'pm25': df_raw['pm25'].mean()
}

# Passing the dictionary to fillna() applies the specific mean to the specific column
df_filled_all = df_raw.fillna(value=fill_values)

print("\nAfter Filling Multiple Columns with Their Respective Means:")
display(df_filled_all)
```

#### **34. How do you detect missing values?**
Pandas provides two identical methods—`isna()` and `isnull()`—to detect missing data. They return a boolean mask of the exact same dimensions as your DataFrame, where `True` indicates a missing value (`NaN` or `None`) and `False` indicates valid data.

**Common Detection Techniques**  
* **The Mask (`df.isna()`):** Best for visual inspection on small datasets or as a foundational step for conditional filtering.
* **Column Totals (`df.isna().sum()`):** The most common diagnostic step. It sums the `True` values (which equal 1) to give you an exact count of missing records per column.
* **Boolean Check (`df.isna().any()`):** Returns a simple True/False per column, answering "Does this column contain *at least one* missing value?"

```python
import pandas as pd
import numpy as np

# Creating a sample dataset with missing values
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town', 'Park Street'],
    'aqi': [115, np.nan, 92, np.nan, 105],
    'pm25': [48.2, 45.0, np.nan, 38.1, 42.1],
    'is_active': [True, True, True, False, np.nan]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. The Boolean Mask
# df.isnull() behaves exactly the same way.
print("\nBoolean Mask of Missing Values:")
display(df.isna())

# 2. Total Missing Values per Column (Most Common)
# Sums the True values vertically down each column.
missing_counts = df.isna().sum()
print("\nMissing values count per column:")
print(missing_counts)

# 3. Quick Boolean Check
# Useful for writing automated checks in data pipelines.
has_missing = df.isna().any()
print("\nDoes the column contain ANY missing values?")
print(has_missing)

# 4. Total Missing Values in the Entire DataFrame
# Chaining sum() twice adds up all the column totals.
total_missing = df.isna().sum().sum()
print(f"\nTotal missing data points in entire dataset: {total_missing}")
```

#### **35. Difference between isna() and notna()?**
`isna()` and `notna()` are exact logical opposites. They both return a boolean mask of your dataset, but they invert which elements evaluate to `True`.

* **`isna()`:** Returns `True` if the data is **missing** (`NaN` or `None`). It is primarily used to count gaps or to extract the broken rows for inspection.
* **`notna()`:** Returns `True` if the data is **valid**. It is primarily used inside brackets for conditional filtering to temporarily exclude missing values from a calculation without permanently dropping them.

```python
import pandas as pd
import numpy as np

# Creating a sample dataset with missing sensor readings
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town'],
    'aqi': [115, np.nan, 92, np.nan]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Comparing the Masks
# Notice how the True/False values are perfectly inverted
print("\nisna() Mask (True = Missing):")
display(df.isna())

print("\nnotna() Mask (True = Valid):")
display(df.notna())

# 2. Using isna() to Inspect Bad Data
# Extracts only the rows where the sensor failed
failed_sensors = df[df['aqi'].isna()]
print("\nIsolating rows WITH missing data using isna():")
display(failed_sensors)

# 3. Using notna() to Filter for Good Data
# Extracts only the rows with valid readings for analysis
# (This is functionally equivalent to df.dropna(subset=['aqi']))
valid_data = df[df['aqi'].notna()]
print("\nIsolating rows with VALID data using notna():")
display(valid_data)
```

You can add a new column to a Pandas DataFrame using standard bracket assignment, mathematical derivation from existing columns, or specialized methods like `insert()` and `assign()`. By default, newly assigned columns are appended to the far right of the DataFrame.

**Common Column Creation Methods**  
* **Bracket Assignment (`df['Col'] = ...`)**: The standard approach. It modifies the DataFrame in place and is used to add static values, lists, or calculated mathematical series.
* **The `insert()` Method**: Allows you to specify exactly where (at which integer index position) the new column should appear, rather than appending it to the end.
* **The `assign()` Method**: Returns a *new* DataFrame with the added column, making it ideal for method chaining because it leaves the original dataset untouched.

```python
import pandas as pd

# Creating a standard dataset
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town'],
    'aqi': [115, 110, 92, 85],
    'pm25': [48.2, 45.0, 40.5, 38.1]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Direct Assignment (Scalar Value)
# Broadcasts a single static value across every row in the new column
df['city'] = 'Kolkata'
print("\nAdded a static 'city' column:")
display(df)

# 2. Direct Assignment (List of Values)
# The list MUST be the exact same length as the DataFrame
df['zone'] = ['West', 'North', 'East', 'East']
print("\nAdded a 'zone' column from a list:")
display(df)

# 3. Derived / Calculated Column
# Creating a new column by applying mathematical operations to existing columns
df['pm25_ratio'] = df['pm25'] / df['aqi']
print("\nAdded a calculated 'pm25_ratio' column:")
display(df)

# 4. The insert() Method
# Syntax: df.insert(loc, column_name, value)
# 'loc=1' places the new column immediately after 'station' (which is at index 0)
df.insert(1, 'sensor_id', [101, 102, 103, 104])
print("\nInserted 'sensor_id' at specific position (index 1):")
display(df)

# 5. The assign() Method (Method Chaining)
# This does NOT modify the original 'df'. It returns a brand new DataFrame.
df_chained = df.assign(
    is_hazardous = df['aqi'] > 100,
    aqi_doubled = df['aqi'] * 2
)
print("\nUsing assign() to create multiple columns simultaneously:")
display(df_chained)
```

#### **37. How do you delete a column?**
The `drop()` method is the primary tool for removing columns from a dataset. By default, it returns a new DataFrame with the specified columns removed, leaving your original dataset entirely intact unless you explicitly pass the parameter `inplace=True`.

**Column Deletion Methods**  
* **`drop(columns=[...])`:** The safest and most readable Pandas method. It clearly indicates you are targeting columns rather than rows.
* **`drop([names], axis=1)`:** The traditional syntax. It requires explicitly setting `axis=1` (columns) because `axis=0` (rows) is the default.
* **`del df['Column']`:** A built-in Python command that deletes the column *in place* immediately. It is fast, but it cannot drop multiple columns at once and cannot be used in method chaining.
* **`pop('Column')`:** Removes the column in place but also returns it as a Series. This is highly useful when you need to extract a target variable for machine learning and remove it from the training dataset in one step.

```python
import pandas as pd

# Creating a dataset with extra, unnecessary columns
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'New Town'],
    'aqi': [115, 110, 92, 85],
    'pm25': [48.2, 45.0, 40.5, 38.1],
    'temp_c': [32, 33, 31, 31],
    'technician': ['Amit', 'Amit', 'Raj', 'Raj']
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Using the 'columns' parameter (Recommended)
# Safely drops a single column and returns a new DataFrame
df_dropped_single = df.drop(columns=['technician'])
print("\nDropped 'technician' column:")
display(df_dropped_single)

# 2. Dropping multiple columns using axis=1
# Passing a list of names and specifying the column axis
df_dropped_multi = df.drop(['temp_c', 'technician'], axis=1)
print("\nDropped multiple columns ('temp_c' and 'technician'):")
display(df_dropped_multi)

# 3. In-place deletion using 'del'
# This modifies 'df' instantly without needing assignment.
del df['technician']
print("\nAfter 'del df[\"technician\"]' (Modifies original):")
display(df)

# 4. Extracting and deleting simultaneously with pop()
# Removes 'pm25' from the DataFrame and saves it to the 'target_variable' Series
target_variable = df.pop('pm25')
print("\nExtracted 'pm25' Series using pop():")
print(target_variable)
print("\nDataFrame after pop() (Notice 'pm25' is gone):")
display(df)
```

#### **39. How do you change data types?**
The `.astype()` method is the standard way to change (cast) data types in Pandas. For more complex conversions involving date formats or messy string data containing text, Pandas provides specialized `to_datetime()` and `to_numeric()` functions.

**Key Type Conversion Methods**  
* **`astype()`:** The standard tool for clean data. Used to convert floats to integers, integers to strings, or strings to memory-efficient categoricals. It will fail if it encounters data it doesn't know how to translate (like converting the string "missing" to an integer).
* **`pd.to_numeric()`:** Converts strings to numeric types (float/int). Crucially, it includes an `errors='coerce'` parameter that forces unparseable text into `NaN` instead of throwing an error.
* **`pd.to_datetime()`:** Intelligently parses date strings (like "2026-09-12" or "12/09/2026") into robust Pandas Timestamp objects, unlocking time-series capabilities.

#### **40. What is astype()?**
The `.astype()` method is the fundamental Pandas function for explicitly casting a Series or an entire DataFrame from one data type to another. It is a strict operation, meaning it will immediately throw a `ValueError` if it encounters even a single data point that cannot be cleanly converted into the target type.

**Key Capabilities**  
* **Dictionary Mapping:** You can pass a dictionary to `.astype()` to cast multiple columns to entirely different data types simultaneously in a single line of code.
* **Memory Optimization:** It is heavily used to downcast data (e.g., converting a 64-bit integer to a 32-bit integer) or to convert repetitive text strings (`object`) into memory-efficient `category` types.
* **Boolean to Integer Casting:** It natively translates boolean masks (`True` / `False`) into binary integers (`1` / `0`), a mandatory step before feeding categorical data into many machine learning algorithms.
* **Not In-Place:** It always returns a new Pandas object. You must reassign the result to overwrite your existing data.

```python
import pandas as pd

# Creating a dataset with inefficient or incorrect data types
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake', 'Howrah', 'Bally'], # object
    'aqi': [115.0, 110.0, 92.0, 120.0, 105.0],                      # float64
    'is_active': [True, True, False, True, False]                   # bool
}
df = pd.DataFrame(data)

print("Original Data Types:\n", df.dtypes)

# 1. Single Column Casting
# Converting AQI from a float to an integer (drops decimals)
df['aqi'] = df['aqi'].astype(int)
print("\nAfter casting 'aqi' to int:\n", df['aqi'].head(2))

# 2. Boolean to Integer Casting
# Translates True to 1 and False to 0
df['is_active'] = df['is_active'].astype(int)
print("\nAfter casting 'is_active' to int:\n", df['is_active'].head(2))

# 3. Multi-Column Casting with a Dictionary
# Resetting the dataframe for the demonstration
df_raw = pd.DataFrame(data)

# Pass a dictionary specifying exactly which column gets which type
type_mapping = {
    'station': 'category', # Great for repeating text strings
    'aqi': 'int32',        # Downcasting to save memory
    'is_active': 'int8'    # Booleans only need the smallest integer type
}

df_optimized = df_raw.astype(type_mapping)

print("\nOptimized Data Types using Dictionary:\n", df_optimized.dtypes)

# 4. The Strictness of astype()
# If we added a string to a numeric column:
try:
    df['aqi'] = [115.0, 110.0, 'Sensor Error', 120.0, 105.0]
    df['aqi'].astype(int)  # This would trigger a ValueError immediately.
except Exception as e:
    print("\n", e)
```

#### **41. How do you apply a function to a column?**
The `.apply()` method is the most versatile tool for executing custom operations on a column. It takes a function—whether it is a built-in Python function, a custom `def` block, or a one-line `lambda` expression—and runs it individually on every single element within that Series.

**Key Application Methods**  
* **Custom Functions (`def`):** Best for complex logic that involves multiple `if/else` conditions (like categorizing AQI numbers into risk levels).
* **Lambda Functions (`lambda`):** Best for quick, simple transformations that only require a single line of code, preventing you from cluttering your script with tiny functions.
* **Built-in Functions:** You can pass standard Python functions like `len`, `str`, or `type` directly into `.apply()` without calling them (e.g., `apply(len)`, not `apply(len())`).

*Note on performance: While extremely flexible, `.apply()` is effectively a standard Python `for` loop under the hood. For simple math, standard Pandas vectorized operations (like `df['aqi'] * 2`) are significantly faster.*

```python
import pandas as pd

# Creating a dataset with raw air quality data
data = {
    'station': ['howrah', 'bally', 'salt lake', 'new town', 'park street'],
    'aqi': [115, 85, 210, 45, 160],
    'pm25': [48.2, 38.1, 110.5, 22.0, 65.0]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Applying a Custom Function (Categorization)
def categorize_aqi(value):
    """Converts a numeric AQI value into a health risk category."""
    if pd.isna(value):
        return 'Unknown'
    elif value <= 50:
        return 'Good'
    elif value <= 100:
        return 'Satisfactory'
    elif value <= 200:
        return 'Moderate'
    else:
        return 'Poor'

# Pass the name of the function WITHOUT parentheses
df['health_risk'] = df['aqi'].apply(categorize_aqi)
print("\nAfter Applying Custom Function (categorize_aqi):")
display(df[['station', 'aqi', 'health_risk']])

# 2. Applying a Lambda Function (Quick Formatting)
# Capitalizes the first letter of each word in the station name
df['station_clean'] = df['station'].apply(lambda x: x.title())

# 3. Applying a Built-in Function
# Calculates the number of characters in each station name
df['name_length'] = df['station'].apply(len)

print("\nAfter Applying Lambda and Built-in Functions:")
display(df[['station_clean', 'name_length']])
```

#### **42. What is apply()?**
The `.apply()` method is an iterator that executes a specified function across an entire Pandas Series or along a specific axis of a DataFrame. It acts as a flexible bridge between custom Python logic and Pandas data structures when built-in methods fall short.

**Mechanics and Behavior**  
* **Execution Strategy:** Under the hood, `.apply()` functions similarly to a standard Python `for` loop. It maps your function to each individual element (if used on a Series) or each row/column (if used on a DataFrame).
* **Axis Control:** When applied to a full DataFrame, the `axis` parameter dictates the direction. `axis=0` (the default) passes entire columns to your function one by one. `axis=1` passes entire rows, allowing you to build complex calculations that use multiple columns simultaneously.
* **The Performance Penalty:** Because it relies on standard Python execution rather than the heavily optimized, C-based architecture of native Pandas methods, `.apply()` is significantly slower. It should be reserved strictly for complex logic (like custom string parsing or multi-variable conditional logic) where standard vectorized math operations cannot be used.

```python
import pandas as pd

# Creating a dataset with multiple pollutant readings
data = {
    'station': ['Howrah', 'Bally', 'Salt Lake'],
    'pm25': [48.2, 65.0, 40.5],
    'pm10': [110.0, 145.2, 85.2]
}
df = pd.DataFrame(data)
print("Original Dataset:")
display(df)

# 1. Applying a function across an entire row (axis=1)
# The function receives the entire row as a Series, allowing us to reference multiple columns.
def calculate_primary_pollutant(row):
    """Determines which pollutant is dominating the air quality profile."""
    if row['pm25'] > (row['pm10'] * 0.5):
        return 'Fine Particulate Heavy (PM2.5)'
    else:
        return 'Coarse Dust Heavy (PM10)'

# By setting axis=1, we process row-by-row instead of column-by-column
df['pollution_profile'] = df.apply(calculate_primary_pollutant, axis=1)

print("\nAfter DataFrame apply() with axis=1:")
display(df)

# 2. Vectorization vs. apply()
df['total_particulates'] = df.apply(lambda row: row['pm25'] + row['pm10'], axis=1) # BAD
df['total_particulates'] = df['pm25'] + df['pm10']  # GOOD # Much faster!
```
