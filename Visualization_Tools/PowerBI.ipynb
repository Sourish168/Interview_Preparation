# 🧠 Important DAX Questions to Practice

For your interview preparation, don't just memorize DAX syntax. Be able to **explain why** each function is used.

For example:

### Total Sales

```DAX
Total Sales =
SUM(Sales[SalesAmount])
```

### Total Profit

```DAX
Total Profit =
SUM(Sales[SalesAmount]) - SUM(Sales[CostAmount])
```

### Profit Margin

```DAX
Profit Margin =
DIVIDE([Total Profit], [Total Sales])
```

### Number of Customers

```DAX
Customers =
DISTINCTCOUNT(Customers[CustomerID])
```

### Average Order Value

```DAX
Average Order Value =
DIVIDE(
    [Total Sales],
    DISTINCTCOUNT(Sales[OrderID])
)
```

The interviewer may then ask:

> **"Why did you use `DIVIDE()` instead of `/`?"**

You should know that `DIVIDE()` provides safer handling of division-by-zero scenarios.
