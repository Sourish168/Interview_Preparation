07. Fibonacci


```python
def fibo(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a+b

fibo(14)
```

08. Factorial


```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n*fact(n-1)

fact(5)
```

09. Prime number


```python
def prime(n):
    if n<=1:
        return False
    elif n==2:
        return True
    else:
        for i in range(2, int(n**0.5)+1):
            if n%i==0:
                return True

    return False

prime(91)
```

10. Anagram


```python
a, b = 'god', 'Dog'

def is_anagram(a, b):
    if len(a)!=len(b):
        return False
    else:
        return sorted(a.upper()) == sorted(b.upper())

is_anagram(a, b)
```

11. Two Sum


```python
ls = [1, 3, 4, 9, 12, 30, 2, 11]
target = 21
def two_sum(ls, target):
    for i in ls:
        if target-i in ls:
            return True
    return False

two_sum(ls, target)
```

13. Find missing number


```python
def missing(ls):
    n = len(ls) + 1
    return n*(n+1)/2 - sum(ls)

ls = [1, 2, 4, 5, 6]
missing(ls)
```

```python
def missing(ls):
    miss_ls = []
    i = 1
    while True:
        if sum(ls[:i]) + sum(miss_ls) - i*(i+1)/2 > 0:
            miss_ls.append(i)
            i += 1
        if i == max(ls):
            break
    return miss_ls

ls = [1, 2, 4, 6, 7]
missing(ls)
```

16. Find common elements

```python
a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]

def common_elements(ls1, ls2):
    common_set = set(ls1) & set(ls2)
    return list(common_set)

common_elements(a, b)
```

17. Longest common prefix

```python
def longest_common_prefix(strs: list[str]) -> str:
    # 1. Handle empty input edge case
    if not strs:
        return ""
    
    # 2. Iterate through the characters of the first string
    first_word = strs[0]
    for i in range(len(first_word)):
        char = first_word[i]
        
        # 3. Compare with the same position in other strings
        for word in strs[1:]:
            # Check if index is out of bounds or if characters mismatch
            if i == len(word) or word[i] != char:
                return first_word[:i]
                
    return first_word

# Example usage:
if __name__ == "__main__":
    example1 = ["flower", "flow", "flight"]
    example2 = ["dog", "racecar", "car"]
    
    print(f"Prefix 1: '{longest_common_prefix(example1)}'")  # Output: 'fl'
    print(f"Prefix 2: '{longest_common_prefix(example2)}'")  # Output: ''
```
